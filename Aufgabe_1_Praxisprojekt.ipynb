### Bereinigte Verkäufe-Tabelle

```python
# CSV-Datei laden
df = pd.read_csv("verkaeufe.csv")

# Funktion zur Umwandlung von Datumswerten
def parse_datum(value):
    value = str(value).strip()

    # Überprüfung und Umwandlung des Datumsformats: DD.MM.YYYY
    if re.fullmatch(r"\d{2}\.\d{2}\.\d{4}", value):
        parts = value.split(".")
        return f"{parts[2]}-{parts[1]}-{parts[0]}"

    # Überprüfung und Rückgabe im Datumsformat: YYYY-MM-DD (ISO-Format)
    if re.fullmatch(r"\d{4}-\d{2}-\d{2}", value):
        return value

    # Überprüfung auf Datum mit Zeitangabe (YYYY-MM-DD HH:MM:SS oder YYYY-MM-DD HH:MM)
    # Nur der Datumsteil wird extrahiert
    if re.fullmatch(r"\d{4}-\d{2}-\d{2} \d{2}:\d{2}(:\d{2})?", value):
        return value.split(" ")[0]

    # Wenn der Wert keinem bekannten Datumsformat entspricht, unverändert zurückgeben
    return value

# Funktion auf die Spalte 'Bestelldatum' anwenden
df["Bestelldatum"] = df["Bestelldatum"].apply(parse_datum)

# Bereinigte Daten in eine neue CSV-Datei speichern
df.to_csv("verkaeufe_bereinigt.csv", index=False)
```
